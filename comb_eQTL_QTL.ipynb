```python
import pandas as pd
```

```python
# Run each QTL separtly
#QTLs_input = pd.read_excel('QTLs_1002.xlsx')
QTLs_input = pd.read_excel('QTLs_1003.xlsx')
QTLs_input.columns.values[0] = 'snp'
```

```python
eQTLs_myeloid = pd.read_excel('eQTLs_myeloid.xlsx', index_col=0)
eQTLs_liver = pd.read_excel('eQTLs_liver.xlsx', index_col=0)
```

```python
myleloid_genes = pd.read_excel('preprocess_sub/myeloid_pos_preprocessed.xlsx', index_col=0)
myleloid_genes = myleloid_genes.iloc[:, [0, 1]]
liver_genes = pd.read_excel('preprocess_sub/liver_pos_preprocessed.xlsx', index_col=0)
liver_genes = liver_genes.iloc[:, [0, 1]]
```

```python
myeloid_gene_pos_dict = myleloid_genes.iloc[:,1].to_dict()
myeloid_gene_chr_dict = myleloid_genes.iloc[:,0].to_dict()
QTLs_input['myeloid_gene_pos'] = QTLs_input['snp'].map(myeloid_gene_pos_dict)
eQTLs_myeloid['gene_pos'] = eQTLs_myeloid['snp'].map(myeloid_gene_pos_dict)

liver_gene_pos_dict = liver_genes.iloc[:,1].to_dict()
liver_gene_chr_dict = liver_genes.iloc[:,0].to_dict()
QTLs_input['liver_gene_pos'] = QTLs_input['snp'].map(liver_gene_pos_dict)
eQTLs_liver['gene_pos'] = eQTLs_liver['snp'].map(liver_gene_pos_dict)
```

```python
distance = 2000000
```

```python
filtered_myeloid__dict = dict()
# check distances between snp to gene
for eqtl in eQTLs_myeloid.iterrows():
   for qtl in QTLs_input.iterrows():
       if abs(qtl[1]['myeloid_gene_pos'] - eqtl[1]['gene_pos']) <= distance:
           filtered_myeloid__dict[eqtl[1].snp] =  qtl[1].snp
```

```python
filtered_liver__dict = dict()
# check distances between snp to gene
for eqtl in eQTLs_liver.iterrows():
    for qtl in QTLs_input.iterrows():
        if abs(qtl[1]['liver_gene_pos'] - eqtl[1]['gene_pos']) <= distance:
                        filtered_liver__dict[eqtl[1].snp] =  qtl[1].snp
```

```python
eQTLs_myeloid['snp_locus'] = eQTLs_myeloid['snp'].map(filtered_myeloid__dict)
intersection_myeloid = eQTLs_myeloid.dropna(subset=['snp_locus'])
```

```python
eQTLs_liver['snp_locus'] = eQTLs_liver['snp'].map(filtered_liver__dict)
intersection_liver = eQTLs_liver.dropna(subset=['snp_locus'])
```

```python
# Run each intersection separtly
# intersection = intersection_myeloid
intersection = intersection_liver
intersection
```

```python
intersection = intersection.loc[:,['gene', 'snp_locus', 'cis/trans']]
```

```python
#intersection.loc[:, 'snp_chr'] = intersection['snp_locus'].map(myeloid_gene_chr_dict)
intersection.loc[:, 'snp_chr'] = intersection['snp_locus'].map(liver_gene_chr_dict)
```

```python
intersection.columns = [col if col != 'snp_locus' else 'snp' for col in intersection.columns]
```

```python
intersection.reset_index(inplace=True)
intersection.drop('index', inplace=True, axis=1)
```
